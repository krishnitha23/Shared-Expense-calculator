```python
rent = int(input("ENTER YOUR ROOM RENT : "))
```

```python
food = int(input("ENTER AMOUNT FOOD ORDERED : "))
```

```python
electricity_spent = float(input("ENTER THE TOTAL ELECTRICITY SPENT : "))
```

```python
charge_per_unitcell = int(input("ENTER THE CHARGE PER UNITCELL : "))
```

```python
persons = int(input("ENTER THE NUMBER OF PERSONS LIVING IN THE ROOM/FLAT :  "))
```

```python
total_bill = electricity_spent * charge_per_unitcell
```

```python
output = (food + rent + total_bill) // persons
```

```python
print ("EACH PERSON WILL PAY : ",output)
```
